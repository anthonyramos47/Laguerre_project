# PH-mesh viewer

Shows the meshes in `data/ph_results/` (planar hexagonal meshes from PH-CPF) in the viewer;
choose them in `mesh_names`. Optionally also their Gauss image.

**Viewer:** `ws.init()` starts a kayviz viewer and prints its URL; open it in a browser. To share one viewer between several notebooks, run `python -m kayviz` in a terminal first.

## 0 — Python path

Add the project root so `LMesh` and `Lconjugacy` import (hanan and kayviz are installed). Adjust `PROJECT_ROOT` if the notebook is moved.

```python
import sys, os


PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
for sub in ("",):  # project root: LMesh/, Lconjugacy/ (hanan, kayviz are installed)
    p = os.path.join(PROJECT_ROOT, sub) if sub else PROJECT_ROOT
    if p not in sys.path:
        sys.path.insert(0, p)

print("Project root:", PROJECT_ROOT)
from kayviz.webscope import script as ws

ws.init()   # pings the server; raises if it is not running
```

```python
import numpy as np
from hanan.geometry.utils import read_obj, normalize_vertices, read_off
import pymeshlab



from hanan.geometry.mesh import Mesh

DATA = os.path.join(PROJECT_ROOT, "data", "ph_results")

mesh_names = [
    "aspiralBdef",
    #"ph_input_cmcOptimized",
    "ph_input_nautilus_ex2",
    #"ph_input_nautilus_channel",
    #"ph_input_seashell_ex5",
    #"ph_input_seashell_ex6",
    #"ph_input_spiral_def",
    #"ph_input_spiral_defM",
    #"ph_input_testSimple"
]

mesh_name = mesh_names[1]  # Change this index to select a different mesh

V, F = read_obj(os.path.join(DATA, mesh_name+".obj"))

mesh = Mesh()
mesh.make_mesh(V, F)



ws.clear()

ws.register_surface_mesh("Original", V, F, color=(0.8, 0.8, 0.8), opacity=0.5)



#sufixes = [".out_d_p_tri.off",".out_d_p.off", ".out_d.off", ".out.off"]
sufixes = [".out_d_p.off", ".out.off"]

mesh_names = [mesh_name+suf for suf in sufixes]

colors_mesh = {"hex":(1,0,0), "tri":(0,0,1)}

for mesh_name in mesh_names:

    V, F = read_off(os.path.join(DATA, mesh_name))

    
    

    if "out_d_p" in mesh_name: 
      
   
      mesh_color =  colors_mesh["hex"]
    else:

      F = [f for f in F if len(f) == 3]  # Filter out non-triangular faces


      mesh_color =  colors_mesh["tri"]

      mesh = Mesh()
      mesh.make_mesh(V, F)
         
      gv, gf, ge = mesh.gauss_image()
      gv = gv + [0,0, 2]
      ws.register_surface_mesh("Gauss Image", gv, gf, color=(0.8, 0.8, 0.8), opacity=0.1, show_edges=True, edge_color=(1.0, 0.5, 0.0), edge_radius = 0.001)


    ws.register_surface_mesh(mesh_name, V, F, color=mesh_color, opacity=0.5,show_edges=True, edge_color=(1.0, 1.0, 1.0), edge_radius = 0.001)
    
       
```
